```python
csv_source = {
    "aleph": [
        "epBooksPre1700", "epBooks1700s", "epBooks1800s_1", "epBooks1800s_2", "epBooks1800s_3",
        "epBooks1800s_4", "Maps", "Music"
    ], 
    "iams": [
        "India Office_v2", "Map Collections_v2", "Music Collections_v2", "Oriental Manuscripts_v2",
        "Philatelic Collections_v2", "Printed Collections_v2", "Qatar_v2", "Sound Archive_v2",
        "Visual Arts_v2", "Western Manuscripts_v2"
    ]
}
```

```python
aleph_dfs = {}

for f in csv_source["aleph"][:1]:
    df = pd.read_csv(f"../data/interim/{f}.csv", encoding="utf8", dtype=dtypes)
    aleph_dfs[f] = df
```

```python
iams_dfs = {}
for f in csv_source["iams"]:
    df = pd.read_csv(f"../data/interim/{f}.csv", encoding="utf8")
    iams_dfs[f] = df
```

```python
for name, df in iams_dfs.items():
    print(f"{name} Reference unique: {df['Record ID'].is_unique}")
```
